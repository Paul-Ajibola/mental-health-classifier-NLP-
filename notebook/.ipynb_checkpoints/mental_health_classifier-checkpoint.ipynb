**Text Preprocessing**

```python
# Lowercasing

def convert_lowercase(text):
    text = text.lower()
    return text 

data['text'] = data['text'].apply(convert_lowercase)
```

```python
# Remove url
def remove_url(text):
    re_url = re.compile('https?.//\S+|www\.\S+')
    return re_url.sub('', text)


data['text'] = data['text'].apply(remove_url)
```

```python
# Removing punctuations
exclude = string.punctuation

def remove_punc(text):
    return text.translate(str.maketrans('', '', exclude))


data['text'] = data['text'].apply(remove_punc)
```

```python
# Remove stopwords

def remove_stopwords(text):
    new_list = []
    words = word_tokenize(text)
    stop_words = set(stopwords.words('english'))   # convert from list to set to optimize search speed
    for word in words:
        if word not in stop_words:
            new_list.append(word)
    return ' '.join(new_list)


data['text'] = data['text'].apply(remove_stopwords)
```

```python
# Stemming 

def perform_stemming(text):
    stemmer = PorterStemmer()
    new_list = []
    words = word_tokenize(text)
    for word in words:
        new_list.append(stemmer.stem(word))

    return " ".join(new_list)

data['text'] = data['text'].apply(perform_stemming)
```

```python
# Creating new column `total words after transformation
data['total words after transformation'] = data['text'].apply(lambda x: np.log(len(x.split())))
```

**Visualize ML Models**

```python
def evaluate_ml_model(model, X_test, y_test, model_name):
    # get predictions and probabilities
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    print(f"\n \--- {model_name} Evaluation Report ---")
    print(classification_report(y_test, y_pred))


    # plot textual report
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='YlGnBu', ax=ax1, cbar=False)
    ax1.set_title(f'{model_name} - COnfusion Matrix')
    ax1.set_xlabel('Predicted Label')
    ax1.set_ylabel('True Label')

    fpr, tpr, _ = roc_curve(y_test, y_prob)  # _ for the threshold
    roc_auc = auc(fpr, tpr)


    ax2.plot(fpr, tpr, color='#26b9ff', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
    ax2.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    ax2.set_xlim([0.0, 1.0])
    ax2.set_ylim([0.0, 1.05])
    ax2.set_title(f'{model_name} - ROC Curve')
    ax2.set_xlabel('False Positive Rate')
    ax2.set_ylabel('True Positive Rate')
    ax2.legend(loc='lower right')
    ax2.grid(True, linestyle='--', alpha=0.6)


    plt.tight_layout()
    plt.show()
```

**RNN**

```python
class RNNClassifier(nn.Module):
    def __init__(self, input_size=2500, hidden_size=32, num_classes=2, num_layers=1, nonlinearity="tanh"):
        super().__init__()
        self.rnn = nn.RNN(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            nonlinearity=nonlinearity,
            batch_first=True,    #(batch, seq, input)
            bidirectional=False
        )
        self.fc = nn.Linear(hidden_size, num_classes)


    def forward(self, x):
        """
        x: (batch, seq_len, input_size)
        returns:
        logits: (batch, num_classes)
        out_seq: (batch, seq_len, hidden_size)   # all hidden states
        h_n: (num_layers, batch, hidden_size)    # final hidden state(s)
        """
        out_seq, h_n = self.rnn(x)
        last_hidden = out_seq[:, -1, :]      # (batch, hidden)
        logits = self.fc(last_hidden)
        return logits, out_seq, h_n
```

```python
rrn_model = RNNClassifier(input_size=2500, hidden_size=64, num_classes=2)
```

**LSTM**

```python
class LSTMClassifier(nn.Module):
    def __init__(self, input_size=2500, hidden_size=64, num_classes=2, num_layers=1, nonlinearity="tanh"):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            # nonlinearity=nonlinearity,  lstm does not recieve tanh arguments
            batch_first=True,    #(batch, seq, input)
            bidirectional=False
        )
        self.fc = nn.Linear(hidden_size, num_classes)


    def forward(self, x):
        """
        x: (batch, seq_len, input_size)
        returns:
        logits: (batch, num_classes)
        out_seq: (batch, seq_len, hidden_size)    # all hidden states
        (h_n, c_n): final hidden state and final cell state
        """
        out_seq, (h_n, c_n) = self.lstm(x)
        last_hidden = out_seq[:, -1, :]   #(batch, hidden)
        logits = self.fc(last_hidden)

        return logits, out_seq, (h_n, c_n)
        
```

```python
lstm_model = LSTMClassifier(input_size=2500, hidden_size=64, num_classes=2)
```

```python
class GRUClassifier(nn.Module):
    def __init__(self, input_size=2500, hidden_size=64, num_classes=2, num_layers=1):
        super().__init__()
        self.gru = nn.GRU(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            bidirectional=False
        )
        self.fc = nn.Linear(hidden_size, num_classes)


    def forward(self, x):
        """
        x: (batch, seq_len, input_size)
        returns:
        logits: (batch, num_classes)
        out_seq: (batch, seq_len, hidden_size)    # all hidden states
        h_n: (num_layers, batch, hidden_size)     # final hidden state
        """
        out_seq, h_n = self.gru(x)

        last_hidden = out_seq[:, -1, :]   # (batch, hidden)
        logits = self.fc(last_hidden)

        return logits, out_seq, h_n
```

```python
gru_model = GRUClassifier(input_size=2500, hidden_size=64, num_classes=2)
```

```python
# design the training function

def train_pytorch_model(model, X_train, y_train, epochs=50, lr=0.001):
    """
    
    """
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model.to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    # prepare data
    X_tensor = torch.tensor(X_train, dtype=torch.float32).unsqueeze(1).to(device)
    y_tensor = torch.tensor(y_train, dtype=torch.long).to(device)

    # create dictionary to collect loss and accuracy values
    history = {"loss": [], "accuracy": []}

    # loop
    for epoch in tqdm(range(epochs), desc="Training Model"):
        model.train()

        # forward pass
        outputs = model(X_tensor)
        logits = outputs[0]
        # calculate the loss
        loss = criterion(logits, y_tensor)
        # clear old memory
        optimizer.zero_grad()
        # backward pass
        loss.backward()
        # update model 
        optimizer.step()

        # calculate probabilities and accuracy
        probabilities = torch.softmax(logits, dim=1)
        predictions = torch.argmax(probabilities, dim=1)

        correct = (predictions == y_tensor).sum().item()
        accuracy = correct / y_tensor.size(0)

        # append loss and accuracy values
        history["loss"].append(loss.item())
        history["accuracy"].append(accuracy)

        tqdm.DataFrame = False
        if (epoch + 1) % 10 == 0:
            tqdm.write(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f} | Accuracy: {accuracy * 100:.2f}")

    return model, history
```

```python
rnn_model, rnn_history = train_pytorch_model(rrn_model, X_train, y_train, epochs=200)
```

```python
lstm_modl, lstm_history = train_pytorch_model(lstm_model, X_train, y_train, epochs=200)
```

```python
gru_model, gru_history = train_pytorch_model(gru_model, X_train, y_train, epochs=200)
```

**Visualization of the DL Models' Results**

```python
def plot_pytorch_history(history, model_name="Model"):
    """
    plots the training loss and accuracy from a pytorch history dictionary.
    """
    epochs = range(1, len(history["loss"]) + 1)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

    # left subplot: loss
    ax1.plot(epochs, history['loss'], color='#6c5ce7', linewidth=2, label='Train Loss')
    ax1.set_title(f"{model_name} - Loss History")
    ax1.set_xlabel('Epochs')
    ax1.set_ylabel('Loss Value')
    ax1.grid(True, linestyle='--', alpha=0.6)
    ax1.legend()


    # left subplot: accuracy
    ax2.plot(epochs, [acc * 100 for acc in history['accuracy']], color='#6c5ce7', linewidth='2', label='Train Accuracy')
    ax2.set_title(f'{model_name} - Accuracy History')
    ax2.set_xlabel('Epochs')
    ax2.set_ylabel('Accuracy (%)')
    ax2.grid(True, linestyle='--', alpha=0.6)
    ax2.legend()

    plt.tight_layout()
    plt.show()
```

```python
plot_pytorch_history(rnn_history, model_name="Vanilla RNN")
```

```python
plot_pytorch_history(lstm_history, model_name="LSTM")
```

```python
plot_pytorch_history(gru_history, model_name="GRU")
```

```python
def evaluate_dl_model(model, X_test, y_test, model_name="PyTorch Model"):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model.eval()


    # prepare test data tensor
    X_test_tensor = torch.tensor(X_test, dtype=torch.float32).unsqueeze(1).to(device)

    # Get predictions and probabilites
    with torch.no_grad():
        outputs = model(X_test_tensor)
        logits = outputs[0]
        probabilities = torch.softmax(logits, dim=1).cpu().numpy()
        predictions = torch.argmax(logits, dim=1).cpu().numpy()


    # print textual metrics summary
    print(f"\n-- {model_name} Evaluation Report ---")
    print(classification_report(y_test, predictions))

    # plot
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # confusion matrix
    cm = confusion_matrix(y_test, predictions)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax1, cbar=False)
    ax1.set_title(f'{model_name} - Confusion Matrix')
    ax1.set_xlabel('Predicted Label')
    ax1.set_ylabel('True Label')


    # ROC Curve
    fpr, tpr, _ = roc_curve(y_test, probabilities[:, 1])
    roc_auc = auc(fpr, tpr)

    ax2.plot(fpr, tpr, color='#ff7675', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
    ax2.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    ax2.set_xlim([0.0, 1.0])
    ax2.set_ylim([0.0, 1.05])
    ax2.set_title(f'{model_name} - ROC Curve')
    ax2.set_xlabel('False Positive Rate')
    ax2.set_ylabel('True Positive Rate')
    ax2.legend(loc='lower right')
    ax2.grid(True, linestyle='--', alpha=0.6)
    

    plt.tight_layout()
    plt.show()
```

```python
evaluate_dl_model(lstm_modl, X_test, y_test, model_name='Long Short Term Memory')
```

```python
evaluate_dl_model(rnn_model, X_test, y_test, model_name='RNN Model')
```

```python
evaluate_dl_model(gru_model, X_test, y_test, model_name='Gated Recurrent Network')
```
